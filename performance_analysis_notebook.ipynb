# Performance Analysis - Degradation Point Detection

This notebook performs comprehensive analysis on performance test data to detect degradation points in system performance. It analyzes response times, resource usage, error rates, and creates interactive visualizations.

````python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px
from datetime import datetime, timedelta
import os
from IPython.display import display, HTML
import warnings
warnings.filterwarnings('ignore')
````

## Helper Functions

````python
def debug_data_structure(csv_data):
    """Function for debugging data structure"""
    print("\nDEBUG DATA STRUCTURE:")
    print(f"Total rows: {len(csv_data)}")
    print(f"Columns: {list(csv_data.columns)}")
    
    if not csv_data.empty:
        print(f"Metric types: {csv_data['metric_name'].unique()}")
        
        # Look at error data
        error_data = csv_data[csv_data['metric_name'] == 'http_req_failed']
        if not error_data.empty:
            print(f"\nHTTP_REQ_FAILED data:")
            print(f"   - Number of records: {len(error_data)}")
            print(f"   - Values: {error_data['metric_value'].unique()[:10]}")
        
        # Look at HTTP request statuses
        http_requests = csv_data[csv_data['metric_name'] == 'http_req_duration']
        if not http_requests.empty and 'status' in http_requests.columns:
            print(f"\nHTTP statuses:")
            print(f"   - Unique statuses: {http_requests['status'].unique()}")
            print(f"   - Example statuses: {http_requests['status'].value_counts().head()}")
        
        # Look at http_reqs data
        http_reqs = csv_data[csv_data['metric_name'] == 'http_reqs']
        if not http_reqs.empty:
            print(f"\nHTTP_REQS data:")
            print(f"   - Number of records: {len(http_reqs)}")
            print(f"   - Values: {http_reqs['metric_value'].unique()[:10]}")
            if 'timestamp' in http_reqs.columns:
                print(f"   - Timestamps: {http_reqs['timestamp'].head(3)}")
````

````python
def calculate_error_rate(csv_data, time_window, window_start, window_end):
    """Correctly calculates error percentage for time interval"""
    try:
        # Find all HTTP requests in this interval
        http_requests = csv_data[
            (csv_data['metric_name'] == 'http_req_duration')
        ].copy()
        
        # Convert timestamp for filtering
        http_requests.loc[:, 'timestamp_dt'] = pd.to_datetime(http_requests['timestamp'], unit='s', errors='coerce')
        http_requests = http_requests.dropna(subset=['timestamp_dt'])
        
        # Filter by time window
        window_requests = http_requests[
            (http_requests['timestamp_dt'] >= window_start) & 
            (http_requests['timestamp_dt'] < window_end)
        ]
        
        total_requests = len(window_requests)
        
        if total_requests == 0:
            return 0, 0
        
        # Count errors by status (4xx, 5xx) or presence of error
        error_count = 0
        
        if 'status' in window_requests.columns:
            # Convert statuses to numeric format without warnings
            status_series = pd.to_numeric(window_requests['status'], errors='coerce')
            error_requests = window_requests[
                (status_series >= 400) |  # HTTP errors
                (window_requests['error'].notna())  # Has error text
            ]
            error_count = len(error_requests)
        
        error_rate = (error_count / total_requests * 300) if total_requests > 0 else 0
        
        return error_count, error_rate
        
    except Exception as e:
        print(f"Error calculating errors: {e}")
        return 0, 0
````

## Degradation Detection Functions

````python
def find_degradation_point(csv_data):
    """Finds degradation point in performance based on response time and errors"""
    
    print("Finding degradation point...")
    
    # Filter data
    duration_data = csv_data[csv_data['metric_name'] == 'http_req_duration']
    
    if duration_data.empty:
        print("No response time data")
        return
    
    # Prepare data
    duration_data = duration_data.copy()
    duration_data['metric_value'] = pd.to_numeric(duration_data['metric_value'], errors='coerce')
    duration_data = duration_data.dropna(subset=['metric_value'])
    
    # Group by time intervals
    if 'timestamp' in duration_data.columns:
        duration_data.loc[:, 'timestamp_dt'] = pd.to_datetime(duration_data['timestamp'], unit='s', errors='coerce')
        duration_data = duration_data.dropna(subset=['timestamp_dt'])
        duration_data = duration_data.sort_values('timestamp_dt')
        
        # Create time intervals (every 10 seconds)
        duration_data.loc[:, 'time_window'] = (duration_data['timestamp_dt'] - duration_data['timestamp_dt'].min()).dt.total_seconds() // 10
        
        # Analyze each time interval
        degradation_found_time = False
        degradation_found_errors = False
        degradation_time_time = None
        degradation_time_errors = None
        degradation_load_time = None
        degradation_load_errors = None
        degradation_error_rate = None
        
        print("\nTime interval analysis:")
        print("Time(sec) | Requests | P95(ms) | Errors | Status")
        print("-" * 65)
        
        for time_window in sorted(duration_data['time_window'].unique()):
            window_data = duration_data[duration_data['time_window'] == time_window]
            
            if len(window_data) < 5:  # Minimum 5 requests for statistics
                continue
            
            # Calculate P95 for this interval
            p95 = np.percentile(window_data['metric_value'], 95)
            
            # Assess load (number of requests in interval)
            load = len(window_data)
            
            # Calculate error percentage for this interval
            window_start = duration_data['timestamp_dt'].min() + pd.Timedelta(seconds=time_window*30)
            window_end = window_start + pd.Timedelta(seconds=30)
            
            error_count, error_rate = calculate_error_rate(csv_data, time_window, window_start, window_end)
            
            # Check degradation conditions
            status = "OK"
            degradation_reason = []
            
            if p95 > 500:
                degradation_reason.append("P95 > 500ms")
                if not degradation_found_time:
                    degradation_found_time = True
                    degradation_time_time = time_window * 10
                    degradation_load_time = load
            
            if error_rate > 3.0:
                degradation_reason.append(f"Errors > 0.5% ({error_rate:.1f}%)")
                if not degradation_found_errors:
                    degradation_found_errors = True
                    degradation_time_errors = time_window * 10
                    degradation_load_errors = load
                    degradation_error_rate = error_rate
            
            if degradation_reason:
                status = f"DEGRADATION ({', '.join(degradation_reason)})"
            
            print(f"{time_window * 10:8.0f} | {load:8} | {p95:7.0f} | {error_rate:5.1f}% | {status}")
        
        # Print results
        print(f"\nANALYSIS RESULTS:")
        
        if degradation_found_time:
            print(f"DEGRADATION BY RESPONSE TIME:")
            print(f"   Time: {degradation_time_time} seconds from start")
            print(f"   Load: ~{degradation_load_time} requests/30sec")
            print(f"   P95 exceeded 500ms")
        
        if degradation_found_errors:
            print(f"DEGRADATION BY ERRORS:")
            print(f"   Time: {degradation_time_errors} seconds from start")
            print(f"   Load: ~{degradation_load_errors} requests/30sec")
            print(f"   Error percentage: {degradation_error_rate:.1f}%")
        
        if not degradation_found_time and not degradation_found_errors:
            print(f"No degradation detected")
        elif not degradation_found_errors:
            print(f"No errors detected (all requests successful)")
    
    return degradation_found_time or degradation_found_errors
````

````python
def create_degradation_analysis(csv_data):
    """Creates analysis to find degradation point by response time and errors"""
    
    print("Creating degradation analysis...")
    
    # Filter response time data
    duration_data = csv_data[csv_data['metric_name'] == 'http_req_duration']
    
    if duration_data.empty:
        print("No response time data")
        return
    
    # Prepare data
    duration_data = duration_data.copy()
    duration_data['metric_value'] = pd.to_numeric(duration_data['metric_value'], errors='coerce')
    duration_data = duration_data.dropna(subset=['metric_value'])
    
    if 'timestamp' not in duration_data.columns:
        print("No timestamps for analysis")
        return
    
    # Convert timestamp
    duration_data.loc[:, 'timestamp_dt'] = pd.to_datetime(duration_data['timestamp'], unit='s', errors='coerce')
    duration_data = duration_data.dropna(subset=['timestamp_dt'])
    duration_data = duration_data.sort_values('timestamp_dt')
    
    # Create time intervals (every 10 seconds)
    duration_data.loc[:, 'time_elapsed'] = (duration_data['timestamp_dt'] - duration_data['timestamp_dt'].min()).dt.total_seconds()
    duration_data.loc[:, 'time_window'] = (duration_data['time_elapsed'] // 10).astype(int)
    
    # Analyze each interval
    window_stats = []
    
    for window in sorted(duration_data['time_window'].unique()):
        window_data = duration_data[duration_data['time_window'] == window]
        
        if len(window_data) < 5:  # Minimum 5 requests for statistics
            continue
        
        # Response time statistics
        p50 = np.percentile(window_data['metric_value'], 50)
        p75 = np.percentile(window_data['metric_value'], 75) 
        p90 = np.percentile(window_data['metric_value'], 90)
        p95 = np.percentile(window_data['metric_value'], 95)
        p99 = np.percentile(window_data['metric_value'], 99)
        
        # Error statistics
        window_start = duration_data['timestamp_dt'].min() + pd.Timedelta(seconds=window*30)
        window_end = window_start + pd.Timedelta(seconds=30)
        
        error_count, error_rate = calculate_error_rate(csv_data, window, window_start, window_end)
        
        total_requests = len(window_data)
        
        window_stats.append({
            'time_window': window,
            'time_seconds': window * 10,
            'requests': total_requests,
            'p50': p50,
            'p75': p75, 
            'p90': p90,
            'p95': p95,
            'p99': p99,
            'error_count': error_count,
            'error_rate': error_rate,
            'degraded_time': p95 > 500,  # Degradation by time
            'degraded_errors': error_rate > 0.5  # Degradation by errors
        })
    
    if not window_stats:
        print("Insufficient data for analysis")
        return
    
    stats_df = pd.DataFrame(window_stats)
    
    # Create interactive degradation analysis using Plotly
    fig = make_subplots(
        rows=3, cols=1,
        subplot_titles=['Response Time P95 and Degradation Threshold', 'Error Rate and Degradation Threshold', 'Load (Number of Requests Over Time)'],
        vertical_spacing=0.08
    )
    
    # Plot 1: P95 over time
    fig.add_trace(
        go.Scatter(
            x=stats_df['time_seconds'],
            y=stats_df['p95'],
            mode='lines+markers',
            name='P95 Response Time',
            line=dict(width=3, color='blue'),
            marker=dict(size=6),
            hovertemplate='Time: %{x}s<br>P95: %{y}ms<extra></extra>'
        ),
        row=1, col=1
    )
    
    # Degradation threshold line
    fig.add_hline(y=500, line_dash="dash", line_color="red", 
                  annotation_text="Degradation Threshold (500ms)", 
                  row=1, col=1)
    
    # Plot 2: Error rate over time
    fig.add_trace(
        go.Scatter(
            x=stats_df['time_seconds'],
            y=stats_df['error_rate'],
            mode='lines+markers',
            name='Error Rate',
            line=dict(width=3, color='red'),
            marker=dict(size=6),
            hovertemplate='Time: %{x}s<br>Error Rate: %{y}%<extra></extra>'
        ),
        row=2, col=1
    )
    
    # Error degradation threshold
    fig.add_hline(y=0.5, line_dash="dash", line_color="darkred", 
                  annotation_text="Degradation Threshold (0.5% errors)", 
                  row=2, col=1)
    
    # Plot 3: Request load over time
    fig.add_trace(
        go.Bar(
            x=stats_df['time_seconds'],
            y=stats_df['requests'],
            name='Requests per 10 seconds',
            marker_color='green',
            opacity=0.7,
            hovertemplate='Time: %{x}s<br>Requests: %{y}<extra></extra>'
        ),
        row=3, col=1
    )
    
    # Mark degradation points if detected
    degraded_time_windows = stats_df[stats_df['degraded_time']]
    if not degraded_time_windows.empty:
        first_degradation = degraded_time_windows.iloc[0]
        degradation_time = first_degradation['time_seconds']
        degradation_requests = first_degradation['requests']
        
        # # Add vertical line for degradation point on all subplots
        # fig.add_vline(x=degradation_time, line_dash="dot", line_color="orange", 
        #               annotation_text=f"Degradation Point: {degradation_time}s", 
        #               row="all", col=1)
        
        # # Add a point showing the degradation
        # fig.add_trace(
        #     go.Scatter(
        #         x=[degradation_time],
        #         y=[degradation_requests],
        #         mode='markers',
        #         marker=dict(size=12, color='red', symbol='circle'),
        #         name='Degradation Point',
        #         hovertemplate=f'Requests at degradation: {degradation_requests}<extra></extra>'
        #     ),
        #     row=3, col=1
        # )
    
    # Update layout
    fig.update_layout(
        height=1000,
        title_text="Performance Degradation Analysis",
        showlegend=True
    )
    
    # Update axes labels
    fig.update_xaxes(title_text="Time from start (seconds)", row=3, col=1)
    fig.update_yaxes(title_text="P95 Response Time (ms)", row=1, col=1)
    fig.update_yaxes(title_text="Error Rate (%)", row=2, col=1)
    fig.update_yaxes(title_text="Number of Requests", row=3, col=1)
    
    fig.show()
    
    print(f"Degradation analysis plot displayed in notebook")
    
    # Print detailed report
    print("\n" + "="*70)
    print("DEGRADATION POINT ANALYSIS")
    print("="*70)
    
    # Degradation by response time
    if not degraded_time_windows.empty:
        first_degraded_time = degraded_time_windows.iloc[0]
        print(f"RESPONSE TIME DEGRADATION:")
        print(f"   Time: {first_degraded_time['time_seconds']} seconds from start")
        print(f"   Load: {first_degraded_time['requests']} requests for 30 seconds")
        print(f"   P95 response time: {first_degraded_time['p95']:.0f} ms")
        print(f"   P99 response time: {first_degraded_time['p99']:.0f} ms")
        
        # Find interval before degradation for comparison
        previous_windows = stats_df[stats_df['time_seconds'] < first_degraded_time['time_seconds']]
        if not previous_windows.empty:
            last_good_window = previous_windows.iloc[-1]
            print(f"   COMPARISON:")
            print(f"      Before degradation: P95 = {last_good_window['p95']:.0f} ms")
            print(f"      At degradation: P95 = {first_degraded_time['p95']:.0f} ms")
            print(f"      Deterioration: +{first_degraded_time['p95'] - last_good_window['p95']:.0f} ms")
    
    # Degradation by errors
    degraded_error_windows = stats_df[stats_df['degraded_errors']]
    if not degraded_error_windows.empty:
        first_degraded_errors = degraded_error_windows.iloc[0]
        print(f"ERROR DEGRADATION:")
        print(f"   Time: {first_degraded_errors['time_seconds']} seconds from start")
        print(f"   Load: {first_degraded_errors['requests']} requests for 30 seconds")
        print(f"   Error rate: {first_degraded_errors['error_rate']:.1f}%")
        print(f"   Number of errors: {first_degraded_errors['error_count']}")
    
    if not degraded_time_windows.empty and not degraded_error_windows.empty:
        # Which degradation happened first
        first_degradation = min(
            degraded_time_windows.iloc[0]['time_seconds'] if not degraded_time_windows.empty else float('inf'),
            degraded_error_windows.iloc[0]['time_seconds'] if not degraded_error_windows.empty else float('inf')
        )
        
        if first_degradation == degraded_time_windows.iloc[0]['time_seconds']:
            print(f"\nFIRST DEGRADATION: by response time")
        else:
            print(f"\nFIRST DEGRADATION: by errors")
    
    if not degraded_time_windows.empty or not degraded_error_windows.empty:
        print(f"\nCONCLUSION: System starts degrading at ~{max(stats_df['requests'])} requests/30sec")
    else:
        print("No degradation detected")
        best_window = stats_df.loc[stats_df['requests'].idxmax()]
        print(f"   Maximum load: {best_window['requests']} requests for 30 seconds")
        print(f"   P95 at max load: {best_window['p95']:.0f} ms")
        print(f"   Errors at max load: {best_window['error_rate']:.1f}%")
    
    print(f"\nOVERALL STATISTICS:")
    print(f"   Intervals analyzed: {len(stats_df)}")
    print(f"   Total requests: {len(duration_data):,}")
    print(f"   Total errors: {sum(stats_df['error_count'])}")
    print(f"   Average P95: {stats_df['p95'].mean():.0f} ms")
    print(f"   Maximum P95: {stats_df['p95'].max():.0f} ms")
    print(f"   Maximum error rate: {stats_df['error_rate'].max():.1f}%")
    
    return stats_df  # Return the dataframe for further analysis if needed
````

````python
def create_percentiles_chart(csv_data):
    """Creates interactive chart with 5 percentile functions over test time"""
    
    print("Creating chart with 5 percentile functions...")
    
    duration_data = csv_data[csv_data['metric_name'] == 'http_req_duration']
    if duration_data.empty:
        print("No response time data")
        return
    
    # Prepare data
    duration_data = duration_data.copy()
    duration_data['metric_value'] = pd.to_numeric(duration_data['metric_value'], errors='coerce')
    duration_data['timestamp'] = pd.to_numeric(duration_data['timestamp'], errors='coerce')
    duration_data = duration_data.dropna(subset=['metric_value', 'timestamp'])
    
    if duration_data.empty:
        return
    
    # Sort by time and create time intervals
    duration_data = duration_data.sort_values('timestamp')
    duration_data['time_interval'] = (duration_data['timestamp'] - duration_data['timestamp'].min())
    
    # Use all available time
    max_time = duration_data['time_interval'].max()
    
    # Break into time intervals
    time_intervals = np.linspace(0, max_time, min(100, int(max_time) + 1))  # adaptive number of points
    percentiles = [50, 75, 90, 95, 99]
    colors = ['green', 'blue', 'orange', 'red', 'purple']
    labels = ['P50', 'P75', 'P90', 'P95', 'P99']
    
    # Calculate percentiles for each time interval
    percentile_over_time = {p: [] for p in percentiles}
    time_points = []
    
    for i in range(1, len(time_intervals)):
        time_start = time_intervals[i-1]
        time_end = time_intervals[i]
        
        # Data for current time interval
        interval_data = duration_data[
            (duration_data['time_interval'] >= time_start) & 
            (duration_data['time_interval'] < time_end)
        ]['metric_value']
        
        if len(interval_data) > 0:
            time_points.append(time_end)  # Use the end time point
            for p in percentiles:
                p_value = np.percentile(interval_data, p)
                percentile_over_time[p].append(p_value)
        else:
            # Add NaN if no data in interval
            time_points.append(time_end)
            for p in percentiles:
                percentile_over_time[p].append(np.nan)
    
    # Create interactive plot using Plotly
    fig = go.Figure()
    
    # Add each percentile as a line
    for p, color, label in zip(percentiles, colors, labels):
        if len(percentile_over_time[p]) > 0:
            fig.add_trace(go.Scatter(
                x=time_points,
                y=percentile_over_time[p],
                mode='lines+markers',
                name=label,
                line=dict(width=2),
                marker=dict(size=4),
                hovertemplate='Time: %{x}s<br>'+label+': %{y}ms<extra></extra>'
            ))
    
    # Add degradation threshold
    fig.add_hline(y=500, line_dash="dash", line_color="black", 
                  annotation_text="Degradation Threshold (500 ms)")
    
    fig.update_layout(
        title='Dynamics of Response Time Percentiles During Test',
        xaxis_title='Test Time (seconds)',
        yaxis_title='Response Time (milliseconds)',
        hovermode='x unified'
    )
    
    fig.show()
    
    print(f"Chart with 5 percentile functions displayed in notebook")
    print(f"Time range: 0-{max_time} seconds")
    
    return percentiles, percentile_over_time
````

````python
def create_cumulative_percentiles_chart(csv_data):
    """Creates interactive chart with cumulative percentiles (all data up to current moment)"""
    
    print("Creating chart with cumulative percentiles...")
    
    duration_data = csv_data[csv_data['metric_name'] == 'http_req_duration']
    if duration_data.empty:
        print("No response time data")
        return
    
    # Prepare data
    duration_data = duration_data.copy()
    duration_data['metric_value'] = pd.to_numeric(duration_data['metric_value'], errors='coerce')
    duration_data['timestamp'] = pd.to_numeric(duration_data['timestamp'], errors='coerce')
    duration_data = duration_data.dropna(subset=['metric_value', 'timestamp'])
    
    if duration_data.empty:
        return
    
    # Sort by time and create time intervals
    duration_data = duration_data.sort_values('timestamp')
    duration_data['time_interval'] = (duration_data['timestamp'] - duration_data['timestamp'].min())
    
    # Use all available time
    max_time = duration_data['time_interval'].max()
    
    # Break into time intervals (increase precision for more data points)
    time_intervals = np.linspace(0, max_time, min(100, int(max_time) + 1))  # adaptive number of points
    percentiles = [50, 75, 90, 95, 99]
    colors = ['green', 'blue', 'orange', 'red', 'purple']
    labels = ['P50', 'P75', 'P90', 'P95', 'P99']
    
    # Calculate CUMULATIVE percentiles for each time interval
    percentile_over_time = {p: [] for p in percentiles}
    time_points = []
    
    for i in range(1, len(time_intervals)):
        time_end = time_intervals[i]
        
        # Take ALL data up to this point in time
        data_so_far = duration_data[duration_data['time_interval'] <= time_end]['metric_value']
        
        if len(data_so_far) > 0:
            time_points.append(time_end)  # Use the end time point
            for p in percentiles:
                p_value = np.percentile(data_so_far, p)
                percentile_over_time[p].append(p_value)
        else:
            # Add NaN if no data up to this point
            time_points.append(time_end)
            for p in percentiles:
                percentile_over_time[p].append(np.nan)
    
    # Create interactive plot using Plotly
    fig = go.Figure()
    
    # Add each cumulative percentile as a line
    for p, color, label in zip(percentiles, colors, labels):
        if len(percentile_over_time[p]) > 0:
            fig.add_trace(go.Scatter(
                x=time_points,
                y=percentile_over_time[p],
                mode='lines+markers',
                name=label,
                line=dict(width=2),
                marker=dict(size=4),
                hovertemplate='Time: %{x}s<br>Cumulative '+label+': %{y}ms<extra></extra>'
            ))
    
    # Add degradation threshold
    fig.add_hline(y=500, line_dash="dash", line_color="black", 
                  annotation_text="Degradation Threshold (500 ms)")
    
    fig.update_layout(
        title='Cumulative Response Time Percentiles (all data up to current moment)',
        xaxis_title='Test Time (seconds)',
        yaxis_title='Response Time (milliseconds)',
        hovermode='x unified'
    )
    
    fig.show()
    
    print(f"Chart with cumulative percentiles displayed in notebook")
    print(f"Time range: 0-{max_time} seconds")
    
    return percentiles, percentile_over_time
````

## Response Time Analysis Functions

````python
def create_response_time_plots(csv_data):
    """Creates interactive response time charts"""
    
    if csv_data.empty:
        print("No k6 data for response time analysis")
        return
    
    print("Response time analysis...")
    
    # Create subplot with 2 charts
    fig = make_subplots(
        rows=2, cols=1,
        subplot_titles=['Response Time by Percentiles (Completed Requests)', 'Response Time Distribution (Completed Requests)'],
        vertical_spacing=0.12
    )
    
    # Plot 1: Response time by percentiles
    try:
        completed_requests = csv_data[csv_data['metric_name'] == 'http_req_duration']
        
        if not completed_requests.empty:
            percentiles = [50, 75, 90, 95, 99]
            percentile_values = [np.percentile(completed_requests['metric_value'], p) for p in percentiles]
            
            fig.add_trace(
                go.Bar(
                    x=[f'P{p}' for p in percentiles],
                    y=percentile_values,
                    name='Response Time by Percentiles',
                    marker_color=['lightgreen', 'lightblue', 'orange', 'red', 'darkred'],
                    hovertemplate='Percentile: %{x}<br>Response Time: %{y}ms<extra></extra>'
                ),
                row=1, col=1
            )
        else:
            fig.add_annotation(
                text="No completed request data",
                xref="x", yref="y",
                x=0.5, y=0.5,
                showarrow=False,
                row=1, col=1
            )
        
    except Exception as e:
        fig.add_annotation(
            text=f'Percentile error: {e}',
            xref="x", yref="y",
            x=0.5, y=0.5,
            showarrow=False,
            row=1, col=1
        )
    
    # Plot 2: Response time histogram
    try:
        completed_requests = csv_data[csv_data['metric_name'] == 'http_req_duration']
        
        if not completed_requests.empty:
            durations = completed_requests['metric_value']
            
            # Calculate statistics
            mean_duration = durations.mean()
            median_duration = durations.median()
            p95_duration = np.percentile(durations, 95)
            
            # Create histogram
            fig.add_trace(
                go.Histogram(
                    x=durations,
                    nbinsx=50,
                    name='Response Time Distribution',
                    marker_color='teal',
                    opacity=0.7,
                    hovertemplate='Response Time: %{x}ms<br>Count: %{y}<extra></extra>'
                ),
                row=2, col=1
            )
            
            # Add vertical lines for statistics
            fig.add_vline(x=mean_duration, line_dash="dash", line_color="red", 
                          annotation_text=f"Mean: {mean_duration:.0f}ms", row=2, col=1)
            fig.add_vline(x=median_duration, line_dash="dash", line_color="green", 
                          annotation_text=f"Median: {median_duration:.0f}ms", row=2, col=1)
            fig.add_vline(x=p95_duration, line_dash="dash", line_color="orange", 
                          annotation_text=f"P95: {p95_duration:.0f}ms", row=2, col=1)
        else:
            fig.add_annotation(
                text="No completed request data",
                xref="x", yref="y",
                x=0.5, y=0.5,
                showarrow=False,
                row=2, col=1
            )
        
    except Exception as e:
        fig.add_annotation(
            text=f'Histogram error: {e}',
            xref="x", yref="y",
            x=0.5, y=0.5,
            showarrow=False,
            row=2, col=1
        )
    
    # Update layout
    fig.update_layout(
        height=800,
        title_text="Interactive Response Time Analysis",
        showlegend=False
    )
    
    # Update axes
    fig.update_xaxes(title_text="Percentile", row=1, col=1)
    fig.update_yaxes(title_text="Response Time (ms)", row=1, col=1)
    fig.update_xaxes(title_text="Response Time (ms)", row=2, col=1)
    fig.update_yaxes(title_text="Number of Requests", row=2, col=1)
    
    fig.show()
    
    print(f"Response time chart displayed in notebook")
````

## Container Resource Analysis Functions

````python
def create_container_plots(resource_data):
    """Creates interactive charts for each container"""
    
    if resource_data.empty:
        print("No resource data for container charts")
        return
    
    containers = resource_data['container'].unique()
    
    for container in containers:
        if container not in ['go_app', 'postgres_db']:
            continue
            
        container_data = resource_data[resource_data['container'] == container]
        if container_data.empty:
            continue
            
        print(f"Creating charts for container: {container}")
        
        # Prepare data with REAL time
        container_data = container_data.copy()
        
        # Convert timestamp to real time (seconds from start of test)
        if 'timestamp' in container_data.columns:
            container_data['timestamp'] = pd.to_numeric(container_data['timestamp'], errors='coerce')
            container_data = container_data.dropna(subset=['timestamp'])
            # Convert to seconds from start of test
            start_time = container_data['timestamp'].min()
            real_time = (container_data['timestamp'] - start_time) / 3000  # in seconds
        else:
            # If no timestamp, use index as approximation
            real_time = range(len(container_data))
        
        # Create CPU usage chart
        try:
            container_data['cpu_numeric'] = container_data['cpu_percent'].str.replace('%', '').astype(float)
            
            # Create interactive chart
            fig = make_subplots(
                rows=2, cols=1,
                subplot_titles=[f'{container} - CPU Usage', f'{container} - Memory Usage'],
                vertical_spacing=0.12
            )
            
            # CPU usage subplot
            fig.add_trace(
                go.Scatter(
                    x=real_time,
                    y=container_data['cpu_numeric'],
                    mode='lines+markers',
                    name='CPU %',
                    line=dict(width=2, color='red'),
                    marker=dict(size=4),
                    hovertemplate='Time: %{x}s<br>CPU: %{y}%<extra></extra>'
                ),
                row=1, col=1
            )
            
            # Calculate and add statistics for CPU
            cpu_mean = container_data['cpu_numeric'].mean()
            cpu_max = container_data['cpu_numeric'].max()
            
            fig.add_hline(y=cpu_mean, line_dash="dash", line_color="blue", 
                          annotation_text=f"Average CPU: {cpu_mean:.1f}%", row=1, col=1)
            fig.add_hline(y=cpu_max, line_dash="dash", line_color="orange", 
                          annotation_text=f"Max CPU: {cpu_max:.1f}%", row=1, col=1)
        
        except Exception as e:
            fig.add_annotation(
                text=f'CPU error: {e}',
                xref="x", yref="y",
                x=0.5, y=0.5,
                showarrow=False,
                row=1, col=1
            )
        
        # Memory usage subplot
        try:
            def parse_memory(mem_str):
                try:
                    used = mem_str.split('/')[0].strip()
                    used_num = float(''.join(filter(lambda x: x.isdigit() or x == '.', used)))
                    return used_num
                except:
                    return 0
            
            container_data['mem_used'] = container_data['mem_usage'].apply(parse_memory)
            
            fig.add_trace(
                go.Scatter(
                    x=real_time,
                    y=container_data['mem_used'],
                    mode='lines+markers',
                    name='Memory (MB)',
                    line=dict(width=2, color='green'),
                    marker=dict(size=4),
                    hovertemplate='Time: %{x}s<br>Memory: %{y} MB<extra></extra>'
                ),
                row=2, col=1
            )
            
            # Calculate and add statistics for memory
            mem_mean = container_data['mem_used'].mean()
            mem_max = container_data['mem_used'].max()
            
            fig.add_hline(y=mem_mean, line_dash="dash", line_color="blue", 
                          annotation_text=f"Average Memory: {mem_mean:.1f} MB", row=2, col=1)
            fig.add_hline(y=mem_max, line_dash="dash", line_color="orange", 
                          annotation_text=f"Max Memory: {mem_max:.1f} MB", row=2, col=1)
            
        except Exception as e:
            fig.add_annotation(
                text=f'Memory error: {e}',
                xref="x", yref="y",
                x=0.5, y=0.5,
                showarrow=False,
                row=2, col=1
            )
        
        # Update layout
        fig.update_layout(
            height=800,
            title_text=f"Interactive Container Analysis: {container}",
            showlegend=False
        )
        
        # Update axes
        fig.update_xaxes(title_text="Test Time (seconds)", row=1, col=1)
        fig.update_yaxes(title_text="CPU %", row=1, col=1)
        fig.update_xaxes(title_text="Test Time (seconds)", row=2, col=1)
        fig.update_yaxes(title_text="Memory (MB)", row=2, col=1)
        
        fig.show()
        
        print(f"Charts for container {container} displayed in notebook")
````

## Main Analysis Function

This function combines all analysis and plotting functions into a comprehensive analysis.

````python
def comprehensive_analysis(csv_data, resource_data=None):
    """Performs comprehensive performance analysis"""
    
    print("=== COMPREHENSIVE PERFORMANCE ANALYSIS ===")
    
    # Debug data structure
    debug_data_structure(csv_data)
    
    # Perform degradation analysis
    print("\n1. Degradation analysis...")
    degradation_result = find_degradation_point(csv_data)
    
    # Create degradation charts
    print("\n2. Creating degradation analysis chart...")
    stats_df = create_degradation_analysis(csv_data)
    
    # Create percentile dynamics charts
    print("\n3. Creating percentile dynamics chart...")
    create_percentiles_chart(csv_data)
    
    # Create cumulative percentiles charts
    print("\n4. Creating cumulative percentiles chart...")
    create_cumulative_percentiles_chart(csv_data)
    
    # Create response time charts
    print("\n5. Creating response time charts...")
    create_response_time_plots(csv_data)
    
    # If resource data is available, create container charts
    if resource_data is not None and not resource_data.empty:
        print("\n6. Creating container resource usage charts...")
        create_container_plots(resource_data)
    
    print("\nCOMPREHENSIVE ANALYSIS COMPLETE!")
    
    return stats_df if 'stats_df' in locals() else None
````
